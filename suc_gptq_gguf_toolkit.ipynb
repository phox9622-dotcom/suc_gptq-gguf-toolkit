```python
#!/bin/bash

# GGUF Model Quantization Script using llama-quantize
# Usage: ./quantize_gguf.sh <model_path> [quantization_types] [--output-dir <path>] [--imatrix <path>]

set -e

# Default quantization types
DEFAULT_QUANT_TYPES="Q2_K Q3_K Q4_K Q5_K Q6_K Q8_K IQ1_S IQ2_XXS IQ2_XS IQ2_S IQ2_M IQ3_XXS IQ3_S IQ3_M IQ4_NL IQ4_XS"

# Default paths
K_QUANT_BINARY="/content/llama-b11721/llama-quantize"
DEFAULT_OUTPUT_DIR="./quantized_models"

print_status() {
    echo -e "[INFO] $1"
}

print_success() {
    echo -e "[SUCCESS] $1"
}

print_warning() {
    echo -e "[WARNING] $1"
}

print_error() {
    echo -e "[ERROR] $1"
}

show_usage() {
    cat << EOF
Usage: $0 <model_path> [quantization_types] [options]

Arguments:
    model_path          Path to the original GGUF model file
    quantization_types  Space-separated list of quantization types (optional)
                       Default: $DEFAULT_QUANT_TYPES

Options:
    --output-dir        Directory where quantized models will be saved
                       Default: ./quantized_models
    --imatrix          Path to the imatrix file for improved quantization quality
    --no-pure          Don't use --pure flag for K-quant quantization
    --llama-quantize   Path to llama-quantize binary
                       Default: $K_QUANT_BINARY

Available quantization types:
    Q2_K, Q3_K, Q4_K, Q5_K, Q6_K, Q8_K
    IQ1_S, IQ2_XXS, IQ2_XS, IQ2_S, IQ2_M
    IQ3_XXS, IQ3_S, IQ3_M, IQ4_NL, IQ4_XS

Examples:
    # Quantize with default types
    $0 model.gguf

    # Quantize with specific types
    $0 model.gguf "Q2_K Q4_K Q6_K"

    # Quantize with imatrix for better quality
    $0 model.gguf "Q2_K Q4_K" --imatrix importance_matrix.dat

    # Custom output directory
    $0 model.gguf --output-dir ./my_quantized_models
EOF
}

validate_quant_type() {
    local quant_type="$1"
    local valid_types="Q2_K Q3_K Q4_K Q5_K Q6_K Q8_K IQ1_S IQ2_XXS IQ2_XS IQ2_S IQ2_M IQ3_XXS IQ3_S IQ3_M IQ4_NL IQ4_XS"

    for valid in $valid_types; do
        if [ "$quant_type" = "$valid" ]; then
            return 0
        fi
    done
    return 1
}

get_file_size() {
    local file="$1"
    if [ -f "$file" ]; then
        du -h "$file" | cut -f1
    else
        echo "N/A"
    fi
}

get_exact_bitwidth() {
    local quant_type="$1"
    case "$quant_type" in
        Q2_K) echo "2.5625" ;;
        Q3_K) echo "3.4375" ;;
        Q4_K) echo "4.5" ;;
        Q5_K) echo "5.5" ;;
        Q6_K) echo "6.5625" ;;
        Q8_K) echo "8.5" ;;
        IQ1_S) echo "1.5625" ;;
        IQ2_XXS) echo "2.0625" ;;
        IQ2_XS) echo "2.3125" ;;
        IQ2_S) echo "2.5" ;;
        IQ2_M) echo "2.7" ;;
        IQ3_XXS) echo "3.0625" ;;
        IQ3_S) echo "3.4375" ;;
        IQ3_M) echo "3.66" ;;
        IQ4_NL) echo "4.25" ;;
        IQ4_XS) echo "4.25" ;;
        *) echo "unknown" ;;
    esac
}

quantize_model() {
    local input_model="$1"
    local output_model="$2"
    local quant_type="$3"
    local imatrix_file="$4"
    local use_pure="$5"

    # Trim whitespace
    quant_type=$(echo "$quant_type" | xargs)

    if [ -f "$output_model" ]; then
        print_warning "Output model already exists: $(basename "$output_model"). Skipping."
        return 0
    fi

    local quant_args=()

    # Add imatrix if provided
    if [ -n "$imatrix_file" ] && [ -f "$imatrix_file" ]; then
        print_status "Using imatrix file: $(basename "$imatrix_file")"
        quant_args+=(--imatrix "$imatrix_file")
    fi

    # Add --pure flag for certain quantization types
    if [ "$use_pure" = true ] && [ "$quant_type" != "Q8_K" ] && [[ ! "$quant_type" =~ ^IQ[12] ]]; then
        print_status "Using --pure option for $quant_type quantization"
        quant_args+=(--pure)
    fi

    quant_args+=("$input_model" "$output_model" "$quant_type")

    print_status "Quantizing to $quant_type ($(get_exact_bitwidth "$quant_type") bits)"
    print_status "Output: $(basename "$output_model")"

    # Execute quantization
    if "$K_QUANT_BINARY" "${quant_args[@]}"; then
        if [ -f "$output_model" ]; then
            local size=$(get_file_size "$output_model")
            print_success "Quantization completed: $quant_type (Size: $size)"
            return 0
        else
            print_error "Quantization completed but output file not found"
            return 1
        fi
    else
        print_error "Failed to quantize to $quant_type"
        return 1
    fi
}

main() {
    if [ $# -lt 1 ]; then
        show_usage
        exit 1
    fi

    local model_path="$1"
    local quant_types="$DEFAULT_QUANT_TYPES"
    local output_dir="$DEFAULT_OUTPUT_DIR"
    local imatrix_file=""
    local use_pure=true

    # Check if model exists
    if [ ! -f "$model_path" ]; then
        print_error "Model file not found: $model_path"
        exit 1
    fi

    shift

    # Parse arguments
    while [ $# -gt 0 ]; do
        case "$1" in
            --output-dir)
                output_dir="$2"
                shift 2
                ;;
            --imatrix)
                imatrix_file="$2"
                shift 2
                ;;
            --no-pure)
                use_pure=false
                shift
                ;;
            --llama-quantize)
                K_QUANT_BINARY="$2"
                shift 2
                ;;
            --help|-h)
                show_usage
                exit 0
                ;;
            -*)
                print_error "Unknown option: $1"
                show_usage
                exit 1
                ;;
            *)
                # Assume it's quantization types
                quant_types="$1"
                shift
                ;;
        esac
    done

    # Validate llama-quantize binary
    if ! command -v "$K_QUANT_BINARY" &> /dev/null; then
        print_error "llama-quantize not found at: $K_QUANT_BINARY"
        print_error "Please install llama.cpp or specify path with --llama-quantize"
        exit 1
    fi

    # Validate imatrix file if provided
    if [ -n "$imatrix_file" ] && [ ! -f "$imatrix_file" ]; then
        print_error "imatrix file not found: $imatrix_file"
        exit 1
    fi

    # Create output directory
    mkdir -p "$output_dir"

    # Get absolute paths
    model_path=$(realpath "$model_path")
    output_dir=$(realpath "$output_dir")
    [ -n "$imatrix_file" ] && imatrix_file=$(realpath "$imatrix_file")

    local model_name=$(basename "$model_path" .gguf)
    local model_size=$(get_file_size "$model_path")

    # Print header
    print_status "=== GGUF Model Quantization ==="
    print_status "Input model: $model_name (Size: $model_size)"
    print_status "Output directory: $output_dir"
    print_status "Quantization types: $quant_types"
    [ -n "$imatrix_file" ] && print_status "imatrix: $(basename "$imatrix_file")"
    echo

    # Test llama-quantize
    print_status "Testing llama-quantize..."
    if "$K_QUANT_BINARY" --help 2>&1 | grep -q "usage"; then
        print_success "llama-quantize is working"
    else
        print_warning "Could not verify llama-quantize. Continuing anyway..."
    fi
    echo

    # Process each quantization type
    local success_count=0
    local total_count=0
    local successful_quants=()

    for quant_type in $quant_types; do
        if ! validate_quant_type "$quant_type"; then
            print_warning "Invalid quantization type: $quant_type (skipping)"
            continue
        fi

        total_count=$((total_count + 1))

        local output_file="$output_dir/${model_name}_${quant_type}.gguf"

        if quantize_model "$model_path" "$output_file" "$quant_type" "$imatrix_file" "$use_pure"; then
            success_count=$((success_count + 1))
            successful_quants+=("$quant_type")
        fi
        echo
    done

    # Create summary JSON
    local summary_file="$output_dir/quantization_summary.json"
    cat > "$summary_file" << EOF
{
  "original_model": "$model_name",
  "original_size": "$model_size",
  "timestamp": "$(date -Iseconds)",
  "quantizations": [
EOF

    local first=true
    for quant in "${successful_quants[@]}"; do
        [ "$first" = false ] && echo "," >> "$summary_file"
        first=false
        local output_file="$output_dir/${model_name}_${quant}.gguf"
        local size=$(get_file_size "$output_file")
        cat >> "$summary_file" << EOF
    {
      "type": "$quant",
      "bitwidth": "$(get_exact_bitwidth "$quant")",
      "file": "$(basename "$output_file")",
      "size": "$size"
    }
EOF
    done

    cat >> "$summary_file" << EOF

  ],
  "imatrix_used": $([ -n "$imatrix_file" ] && echo "true" || echo "false")
}
EOF

    # Print summary
    echo
    print_status "=== QUANTIZATION SUMMARY ==="
    print_status "Successful: $success_count/$total_count"
    print_status "Output directory: $output_dir"

    if [ ${#successful_quants[@]} -gt 0 ]; then
        echo
        print_status "Quantized models:"
        for quant in "${successful_quants[@]}"; do
            local output_file="$output_dir/${model_name}_${quant}.gguf"
            print_status "  • $quant ($(get_exact_bitwidth "$quant") bits): $(get_file_size "$output_file")"
        done
    fi

    if [ $success_count -gt 0 ]; then
        print_success "Quantization completed successfully!"
        print_status "Summary saved to: $summary_file"
    else
        print_error "No quantizations were successful!"
        exit 1
    fi
}

# Run main function
if [ "${BASH_SOURCE[0]}" = "${0}" ]; then
    main "$@"
fi
```

```python


#!/bin/bash

# ==============================================================================
# GGUF Model Quantization Script using llama-quantize
# Optimized for Local & Google Colab environments
# ==============================================================================

set -eo pipefail

# Default configuration
DEFAULT_OUTPUT_DIR="./quantized_models"
DEFAULT_QUANT_TYPES="Q4_K_M Q5_K_M Q8_0"

# Auto-detect CPU cores (Fallback to 2 if undetected)
AUTO_THREADS=$(nproc 2>/dev/null || echo 2)
THREADS="$AUTO_THREADS"

# Flags
ALLOW_REQUANTIZE=false
USE_PURE=false
IMATRIX_FILE=""
OUTPUT_DIR="$DEFAULT_OUTPUT_DIR"
MODEL_PATH=""
QUANT_TYPES=""

# Candidate paths for llama-quantize binary
CANDIDATE_BINARIES=(
    "/content/llama-b11721/llama-quantize"
    "/content/llama.cpp/build/bin/llama-quantize"
    "$(which llama-quantize 2>/dev/null)"
    "./llama-quantize"
)

K_QUANT_BINARY=""
for bin in "${CANDIDATE_BINARIES[@]}"; do
    if [ -n "$bin" ] && [ -x "$bin" ]; then
        K_QUANT_BINARY="$bin"
        break
    fi
done

# ANSI Color codes for clean output
CLR_RESET="\033[0m"
CLR_INFO="\033[1;34m"
CLR_SUCCESS="\033[1;32m"
CLR_WARN="\033[1;33m"
CLR_ERROR="\033[1;31m"

print_status()  { echo -e "${CLR_INFO}[INFO]${CLR_RESET} $1"; }
print_success() { echo -e "${CLR_SUCCESS}[SUCCESS]${CLR_RESET} $1"; }
print_warning() { echo -e "${CLR_WARN}[WARNING]${CLR_RESET} $1"; }
print_error()   { echo -e "${CLR_ERROR}[ERROR]${CLR_RESET} $1"; }

show_usage() {
    cat << EOF
Usage: $0 <model_path> [quantization_types] [options]

Arguments:
    model_path          Path to the input GGUF model file (F16, BF16, or Q8)
    quantization_types  Space or comma-separated list of types (Default: "$DEFAULT_QUANT_TYPES")

Options:
    -o, --output-dir <path>   Directory to save quantized models (Default: $DEFAULT_OUTPUT_DIR)
    -t, --threads <num>       Number of CPU threads to use (Default: auto-detected: $AUTO_THREADS)
    -i, --imatrix <path>      Path to importance matrix (.dat) file
    --allow-requantize        Allow re-quantizing already quantized models (e.g. from Q8_0)
    --pure                    Quantize all tensors to the exact type (Not recommended for small models)
    --llama-quantize <path>   Explicit path to llama-quantize executable
    -h, --help                Show this help message

Supported Types:
    Q2_K, Q3_K_S, Q3_K_M, Q3_K_L, Q4_0, Q4_1, Q4_K_S, Q4_K_M,
    Q5_0, Q5_1, Q5_K_S, Q5_K_M, Q6_K, Q8_0, Q8_K,
    IQ1_S, IQ1_M, IQ2_XXS, IQ2_XS, IQ2_S, IQ2_M, IQ3_XXS, IQ3_S, IQ3_M, IQ4_NL, IQ4_XS

Examples:
    $0 model.gguf "Q4_K_M"
    $0 model.gguf "Q4_K_M,Q5_K_M,Q8_0" --threads 2
    $0 q8_model.gguf "Q4_K_M" --allow-requantize
    $0 model.gguf "Q2_K Q3_K_M" --imatrix imatrix.dat -o ./my_models
EOF
}

validate_quant_type() {
    local q="$1"
    local valid_types="
        Q2_K Q3_K_S Q3_K_M Q3_K_L Q4_0 Q4_1 Q4_K_S Q4_K_M
        Q5_0 Q5_1 Q5_K_S Q5_K_M Q6_K Q8_0 Q8_K
        IQ1_S IQ1_M IQ2_XXS IQ2_XS IQ2_S IQ2_M IQ3_XXS IQ3_S IQ3_M IQ4_NL IQ4_XS
    "
    for v in $valid_types; do
        if [ "$q" = "$v" ]; then
            return 0
        fi
    done
    return 1
}

get_file_size() {
    local file="$1"
    if [ -f "$file" ]; then
        du -h "$file" | cut -f1
    else
        echo "N/A"
    fi
}

get_exact_bitwidth() {
    case "$1" in
        IQ1_S)   echo "1.56" ;;
        IQ1_M)   echo "1.75" ;;
        IQ2_XXS) echo "2.06" ;;
        IQ2_XS)  echo "2.31" ;;
        IQ2_S)   echo "2.50" ;;
        IQ2_M)   echo "2.70" ;;
        Q2_K)    echo "2.62" ;;
        IQ3_XXS) echo "3.06" ;;
        IQ3_S)   echo "3.44" ;;
        IQ3_M)   echo "3.66" ;;
        Q3_K_S)  echo "3.44" ;;
        Q3_K_M)  echo "3.66" ;;
        Q3_K_L)  echo "3.82" ;;
        Q4_0)    echo "4.00" ;;
        Q4_1)    echo "4.25" ;;
        IQ4_NL)  echo "4.25" ;;
        IQ4_XS)  echo "4.25" ;;
        Q4_K_S)  echo "4.45" ;;
        Q4_K_M)  echo "4.50" ;;
        Q4_K)    echo "4.50" ;;
        Q5_0)    echo "5.00" ;;
        Q5_1)    echo "5.25" ;;
        Q5_K_S)  echo "5.35" ;;
        Q5_K_M)  echo "5.50" ;;
        Q5_K)    echo "5.50" ;;
        Q6_K)    echo "6.56" ;;
        Q8_0)    echo "8.00" ;;
        Q8_K)    echo "8.50" ;;
        *)       echo "N/A" ;;
    esac
}

quantize_model() {
    local in_file="$1"
    local out_file="$2"
    local q_type="$3"

    if [ -f "$out_file" ]; then
        print_warning "Output model already exists: $(basename "$out_file"). Skipping."
        return 0
    fi

    local args=()

    # Pass imatrix if available
    if [ -n "$IMATRIX_FILE" ]; then
        args+=(--imatrix "$IMATRIX_FILE")
    fi

    # Allow re-quantization from Q8 / quantized inputs
    if [ "$ALLOW_REQUANTIZE" = true ]; then
        args+=(--allow-requantize)
    fi

    # Pure flag only if explicitly requested
    if [ "$USE_PURE" = true ]; then
        args+=(--pure)
    fi

    # Core arguments: input, output, type, threads
    args+=("$in_file" "$out_file" "$q_type" "$THREADS")

    print_status "Quantizing to ${CLR_SUCCESS}$q_type${CLR_RESET} (~$(get_exact_bitwidth "$q_type") bpw) using $THREADS threads..."
    print_status "Target: $(basename "$out_file")"

    local start_time=$(date +%s)

    if "$K_QUANT_BINARY" "${args[@]}"; then
        local end_time=$(date +%s)
        local elapsed=$((end_time - start_time))
        if [ -f "$out_file" ]; then
            print_success "Finished $q_type in ${elapsed}s | Size: $(get_file_size "$out_file")"
            return 0
        else
            print_error "Quantization finished but file was not created: $out_file"
            return 1
        fi
    else
        print_error "Failed to quantize to $q_type"
        return 1
    fi
}

main() {
    # Argument parsing
    while [ $# -gt 0 ]; do
        case "$1" in
            -h|--help)
                show_usage
                exit 0
                ;;
            -o|--output-dir)
                OUTPUT_DIR="$2"
                shift 2
                ;;
            -t|--threads)
                THREADS="$2"
                shift 2
                ;;
            -i|--imatrix)
                IMATRIX_FILE="$2"
                shift 2
                ;;
            --allow-requantize)
                ALLOW_REQUANTIZE=true
                shift
                ;;
            --pure)
                USE_PURE=true
                shift
                ;;
            --llama-quantize)
                K_QUANT_BINARY="$2"
                shift 2
                ;;
            -*)
                print_error "Unknown option: $1"
                show_usage
                exit 1
                ;;
            *)
                if [ -z "$MODEL_PATH" ]; then
                    MODEL_PATH="$1"
                elif [ -z "$QUANT_TYPES" ]; then
                    QUANT_TYPES="$1"
                else
                    print_error "Unexpected argument: $1"
                    exit 1
                fi
                shift
                ;;
        esac
    done

    # Validation
    if [ -z "$MODEL_PATH" ]; then
        print_error "Model path is required."
        show_usage
        exit 1
    fi

    if [ ! -f "$MODEL_PATH" ]; then
        print_error "Model file not found: $MODEL_PATH"
        exit 1
    fi

    if [ -z "$K_QUANT_BINARY" ] || [ ! -x "$K_QUANT_BINARY" ]; then
        print_error "llama-quantize binary not found or not executable!"
        print_error "Specify it with --llama-quantize /path/to/llama-quantize"
        exit 1
    fi

    if [ -n "$IMATRIX_FILE" ] && [ ! -f "$IMATRIX_FILE" ]; then
        print_error "imatrix file not found: $IMATRIX_FILE"
        exit 1
    fi

    # Set default quantization types if not provided
    if [ -z "$QUANT_TYPES" ]; then
        QUANT_TYPES="$DEFAULT_QUANT_TYPES"
    fi

    # Replace commas with spaces to support both "Q4_K_M,Q5_K_M" and "Q4_K_M Q5_K_M"
    QUANT_TYPES="${QUANT_TYPES//,/ }"

    # Ensure absolute paths
    MODEL_PATH=$(realpath "$MODEL_PATH")
    mkdir -p "$OUTPUT_DIR"
    OUTPUT_DIR=$(realpath "$OUTPUT_DIR")
    [ -n "$IMATRIX_FILE" ] && IMATRIX_FILE=$(realpath "$IMATRIX_FILE")

    local model_name=$(basename "$MODEL_PATH" .gguf)
    local model_size=$(get_file_size "$MODEL_PATH")

    echo
    print_status "================ GGUF QUANTIZATION PIPELINE ================"
    print_status "Input Model     : $model_name ($model_size)"
    print_status "Binary Path     : $K_QUANT_BINARY"
    print_status "CPU Threads     : $THREADS"
    print_status "Output Dir      : $OUTPUT_DIR"
    print_status "Quant Types     : $QUANT_TYPES"
    print_status "Allow Requant   : $ALLOW_REQUANTIZE"
    print_status "Use Pure        : $USE_PURE"
    [ -n "$IMATRIX_FILE" ] && print_status "imatrix File    : $IMATRIX_FILE"
    print_status "============================================================"
    echo

    local successful_quants=()
    local success_count=0
    local total_count=0

    for q_type in $QUANT_TYPES; do
        # Trim whitespace
        q_type=$(echo "$q_type" | xargs)
        [ -z "$q_type" ] && continue

        if ! validate_quant_type "$q_type"; then
            print_warning "Invalid quantization type: '$q_type'. Skipping."
            continue
        fi

        total_count=$((total_count + 1))
        local output_file="$OUTPUT_DIR/${model_name}_${q_type}.gguf"

        if quantize_model "$MODEL_PATH" "$output_file" "$q_type"; then
            success_count=$((success_count + 1))
            successful_quants+=("$q_type")
        fi
        echo
    done

    # Generate JSON summary
    local summary_file="$OUTPUT_DIR/${model_name}_quant_summary.json"
    cat > "$summary_file" << EOF
{
  "original_model": "$model_name",
  "original_size": "$model_size",
  "date_utc": "$(date -u +"%Y-%m-%dT%H:%M:%SZ")",
  "threads_used": $THREADS,
  "imatrix_used": $([ -n "$IMATRIX_FILE" ] && echo "true" || echo "false"),
  "quantizations": [
EOF

    local first=true
    for q in "${successful_quants[@]}"; do
        [ "$first" = false ] && echo "," >> "$summary_file"
        first=false
        local out_f="$OUTPUT_DIR/${model_name}_${q}.gguf"
        cat >> "$summary_file" << EOF
    {
      "type": "$q",
      "approx_bpw": "$(get_exact_bitwidth "$q")",
      "file": "$(basename "$out_f")",
      "size": "$(get_file_size "$out_f")"
    }
EOF
    done

    cat >> "$summary_file" << EOF

  ]
}
EOF

    # Final summary display
    print_status "=================== PROCESS COMPLETED ==================="
    print_status "Total Successful : $success_count / $total_count"
    print_status "Summary JSON     : $summary_file"
    echo
    for q in "${successful_quants[@]}"; do
        local out_f="$OUTPUT_DIR/${model_name}_${q}.gguf"
        echo -e "  • ${CLR_SUCCESS}$q${CLR_RESET} (~$(get_exact_bitwidth "$q") bpw): $(get_file_size "$out_f")"
    done
    print_status "========================================================="
}

if [ "${BASH_SOURCE[0]}" = "${0}" ]; then
    main "$@"
fi
```

```python

/content/q.sh







#!/bin/bash

# GGUF Model Quantization Script using llama-quantize
# Usage: ./quantize_gguf.sh <model_path> [quantization_types] [--output-dir <path>] [--imatrix <path>]

set -e

# Default quantization types (بدء التكميم بصيغ IQ أولاً)
DEFAULT_QUANT_TYPES="IQ1_S IQ2_XXS IQ2_XS IQ2_S IQ2_M IQ3_XXS IQ3_S IQ3_M IQ4_NL IQ4_XS Q2_K Q3_K Q4_K Q5_K Q6_K Q8_K"

# Default paths (يفحص مسار كولاب ثم المسار المحلي)
if [ -f "/content/llama-b11721/llama-quantize" ]; then
    K_QUANT_BINARY="/content/llama-b11721/llama-quantize"
elif [ -f "$HOME/bin/llama/llama-quantize" ]; then
    K_QUANT_BINARY="$HOME/bin/llama/llama-quantize"
else
    K_QUANT_BINARY="llama-quantize"
fi

DEFAULT_OUTPUT_DIR="./quantized_models"

print_status() {
    echo -e "[INFO] $1"
}

print_success() {
    echo -e "[SUCCESS] $1"
}

print_warning() {
    echo -e "[WARNING] $1"
}

print_error() {
    echo -e "[ERROR] $1"
}

show_usage() {
    cat << EOF
Usage: $0 <model_path> [quantization_types] [options]

Arguments:
    model_path          Path to the original GGUF model file
    quantization_types  Space-separated list of quantization types (optional)
                       Default: $DEFAULT_QUANT_TYPES

Options:
    --output-dir        Directory where quantized models will be saved
                       Default: ./quantized_models
    --imatrix          Path to the imatrix file (Required for most IQ quants!)
    --allow-requantize  Allow re-quantizing an already quantized model
    --no-pure          Don't use --pure flag for K-quant quantization
    --llama-quantize   Path to llama-quantize binary
                       Default: $K_QUANT_BINARY

Available quantization types (Starting from IQ):
    IQ1_S, IQ2_XXS, IQ2_XS, IQ2_S, IQ2_M
    IQ3_XXS, IQ3_S, IQ3_M, IQ4_NL, IQ4_XS
    Q2_K, Q3_K, Q4_K, Q5_K, Q6_K, Q8_K

Examples:
    # Quantize with default types (starts from IQ)
    $0 model.gguf

    # Quantize with specific IQ types
    $0 model.gguf "IQ2_XXS IQ3_XXS IQ4_NL" --imatrix imatrix.dat

    # Custom output directory
    $0 model.gguf --output-dir ./my_quantized_models
EOF
}

validate_quant_type() {
    local quant_type="$1"
    local valid_types="IQ1_S IQ2_XXS IQ2_XS IQ2_S IQ2_M IQ3_XXS IQ3_S IQ3_M IQ4_NL IQ4_XS Q2_K Q3_K Q4_K Q5_K Q6_K Q8_K"

    for valid in $valid_types; do
        if [ "$quant_type" = "$valid" ]; then
            return 0
        fi
    done
    return 1
}

get_file_size() {
    local file="$1"
    if [ -f "$file" ]; then
        du -h "$file" | cut -f1
    else
        echo "N/A"
    fi
}

get_exact_bitwidth() {
    local quant_type="$1"
    case "$quant_type" in
        IQ1_S) echo "1.5625" ;;
        IQ2_XXS) echo "2.0625" ;;
        IQ2_XS) echo "2.3125" ;;
        IQ2_S) echo "2.5" ;;
        IQ2_M) echo "2.7" ;;
        IQ3_XXS) echo "3.0625" ;;
        IQ3_S) echo "3.4375" ;;
        IQ3_M) echo "3.66" ;;
        IQ4_NL) echo "4.25" ;;
        IQ4_XS) echo "4.25" ;;
        Q2_K) echo "2.5625" ;;
        Q3_K) echo "3.4375" ;;
        Q4_K) echo "4.5" ;;
        Q5_K) echo "5.5" ;;
        Q6_K) echo "6.5625" ;;
        Q8_K) echo "8.5" ;;
        *) echo "unknown" ;;
    esac
}

quantize_model() {
    local input_model="$1"
    local output_model="$2"
    local quant_type="$3"
    local imatrix_file="$4"
    local use_pure="$5"
    local allow_requantize="$6"

    # Trim whitespace
    quant_type=$(echo "$quant_type" | xargs)

    if [ -f "$output_model" ]; then
        print_warning "Output model already exists: $(basename "$output_model"). Skipping."
        return 0
    fi

    local quant_args=()

    # Add allow-requantize if enabled
    if [ "$allow_requantize" = true ]; then
        quant_args+=(--allow-requantize)
    fi

    # Add imatrix if provided
    if [ -n "$imatrix_file" ] && [ -f "$imatrix_file" ]; then
        print_status "Using imatrix file: $(basename "$imatrix_file")"
        quant_args+=(--imatrix "$imatrix_file")
    fi

    # Add --pure flag for certain quantization types
    if [ "$use_pure" = true ] && [ "$quant_type" != "Q8_K" ] && [[ ! "$quant_type" =~ ^IQ[12] ]]; then
        print_status "Using --pure option for $quant_type quantization"
        quant_args+=(--pure)
    fi

    quant_args+=("$input_model" "$output_model" "$quant_type")

    print_status "Quantizing to $quant_type ($(get_exact_bitwidth "$quant_type") bits)"
    print_status "Output: $(basename "$output_model")"

    # Execute quantization
    if "$K_QUANT_BINARY" "${quant_args[@]}"; then
        if [ -f "$output_model" ]; then
            local size=$(get_file_size "$output_model")
            print_success "Quantization completed: $quant_type (Size: $size)"
            return 0
        else
            print_error "Quantization completed but output file not found"
            return 1
        fi
    else
        print_error "Failed to quantize to $quant_type"
        return 1
    fi
}

main() {
    if [ $# -lt 1 ]; then
        show_usage
        exit 1
    fi

    local model_path="$1"
    local quant_types="$DEFAULT_QUANT_TYPES"
    local output_dir="$DEFAULT_OUTPUT_DIR"
    local imatrix_file=""
    local use_pure=true
    local allow_requantize=false

    # Check if model exists
    if [ ! -f "$model_path" ]; then
        print_error "Model file not found: $model_path"
        exit 1
    fi

    shift

    # Parse arguments
    while [ $# -gt 0 ]; do
        case "$1" in
            --output-dir)
                output_dir="$2"
                shift 2
                ;;
            --imatrix)
                imatrix_file="$2"
                shift 2
                ;;
            --allow-requantize)
                allow_requantize=true
                shift
                ;;
            --no-pure)
                use_pure=false
                shift
                ;;
            --llama-quantize)
                K_QUANT_BINARY="$2"
                shift 2
                ;;
            --help|-h)
                show_usage
                exit 0
                ;;
            -*)
                print_error "Unknown option: $1"
                show_usage
                exit 1
                ;;
            *)
                # Assume it's quantization types
                quant_types="$1"
                shift
                ;;
        esac
    done

    # Validate llama-quantize binary
    if ! command -v "$K_QUANT_BINARY" &> /dev/null && [ ! -x "$K_QUANT_BINARY" ]; then
        print_error "llama-quantize not found at: $K_QUANT_BINARY"
        print_error "Please install llama.cpp or specify path with --llama-quantize"
        exit 1
    fi

    # Validate imatrix file if provided
    if [ -n "$imatrix_file" ] && [ ! -f "$imatrix_file" ]; then
        print_error "imatrix file not found: $imatrix_file"
        exit 1
    fi

    # Create output directory
    mkdir -p "$output_dir"

    # Get absolute paths
    model_path=$(realpath "$model_path")
    output_dir=$(realpath "$output_dir")
    [ -n "$imatrix_file" ] && imatrix_file=$(realpath "$imatrix_file")

    local model_name=$(basename "$model_path" .gguf)
    local model_size=$(get_file_size "$model_path")

    # Print header
    print_status "=== GGUF Model Quantization ==="
    print_status "Input model: $model_name (Size: $model_size)"
    print_status "Output directory: $output_dir"
    print_status "Quantization types: $quant_types"
    [ -n "$imatrix_file" ] && print_status "imatrix: $(basename "$imatrix_file")"
    echo

    # Test llama-quantize
    print_status "Testing llama-quantize..."
    if "$K_QUANT_BINARY" --help 2>&1 | grep -q -i "usage"; then
        print_success "llama-quantize is working"
    else
        print_warning "Could not verify llama-quantize. Continuing anyway..."
    fi
    echo

    # Process each quantization type
    local success_count=0
    local total_count=0
    local successful_quants=()

    for quant_type in $quant_types; do
        if ! validate_quant_type "$quant_type"; then
            print_warning "Invalid quantization type: $quant_type (skipping)"
            continue
        fi

        total_count=$((total_count + 1))

        local output_file="$output_dir/${model_name}_${quant_type}.gguf"

        if quantize_model "$model_path" "$output_file" "$quant_type" "$imatrix_file" "$use_pure" "$allow_requantize"; then
            success_count=$((success_count + 1))
            successful_quants+=("$quant_type")
        fi
        echo
    done

    # Create summary JSON
    local summary_file="$output_dir/quantization_summary.json"
    cat > "$summary_file" << EOF
{
  "original_model": "$model_name",
  "original_size": "$model_size",
  "timestamp": "$(date -u +"%Y-%m-%dT%H:%M:%SZ")",
  "quantizations": [
EOF

    local first=true
    for quant in "${successful_quants[@]}"; do
        [ "$first" = false ] && echo "," >> "$summary_file"
        first=false
        local output_file="$output_dir/${model_name}_${quant}.gguf"
        local size=$(get_file_size "$output_file")
        cat >> "$summary_file" << EOF
    {
      "type": "$quant",
      "bitwidth": "$(get_exact_bitwidth "$quant")",
      "file": "$(basename "$output_file")",
      "size": "$size"
    }
EOF
    done

    cat >> "$summary_file" << EOF

  ],
  "imatrix_used": $([ -n "$imatrix_file" ] && echo "true" || echo "false")
}
EOF

    # Print summary
    echo
    print_status "=== QUANTIZATION SUMMARY ==="
    print_status "Successful: $success_count/$total_count"
    print_status "Output directory: $output_dir"

    if [ ${#successful_quants[@]} -gt 0 ]; then
        echo
        print_status "Quantized models:"
        for quant in "${successful_quants[@]}"; do
            local output_file="$output_dir/${model_name}_${quant}.gguf"
            print_status "  • $quant ($(get_exact_bitwidth "$quant") bits): $(get_file_size "$output_file")"
        done
    fi

    if [ $success_count -gt 0 ]; then
        print_success "Quantization completed successfully!"
        print_status "Summary saved to: $summary_file"
    else
        print_error "No quantizations were successful!"
        exit 1
    fi
}

# Run main function
if [ "${BASH_SOURCE[0]}" = "${0}" ]; then
    main "$@"
fi
```
